```python

import pandas as pd

df = pd.read_csv("istanbul_aqi_data_combined.csv")

print("Veri setinin boyutu:", df.shape)

display(df.head())

df.info()
```

```python

print("Sütun isimleri:")
print(df.columns.tolist())

print("\nEksik değerler:")
print(df.isnull().sum())
```

```python

# Her sütundaki eksik değer sayısı
print(df.isnull().sum())

# Toplam eksik değer sayısı
print("Toplam null:", df.isnull().sum().sum())

# Eksik değer içeren satırları göster
display(df[df.isnull().any(axis=1)])
```

```python

# Timestamp sütununu datetime tipine dönüştür
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Hedef değişkenimiz PM2.5
target = "PM2.5"

print("Hedef değişken:", target)
print("Tarih aralığı:")
print(df["Timestamp"].min())
print(df["Timestamp"].max())

print("\nBölge sayısı:", df["Province"].nunique())
print("\nBölgeler:")
print(df["Province"].unique())

print("\nHedef değişkenin veri tipi:")
print(df[target].dtype)
```
